```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

# Ruta correcta (IMPORTANTE)
ruta = "/content/drive/MyDrive/Datos/2019-Oct.csv"

# Cargar SOLO una parte
df = pd.read_csv(ruta, nrows=100000)

df.head()
```

```python
import matplotlib.pyplot as plt

eventos = df['event_type'].value_counts()

eventos.plot(kind='bar')
plt.title("Cantidad de eventos por tipo")
plt.xlabel("Tipo de evento")
plt.ylabel("Cantidad")
plt.show()
```

```python
df['price'].hist()
plt.title("Distribución de precios")
plt.xlabel("Precio")
plt.ylabel("Frecuencia")
plt.show()
```

```python
df.isnull().sum().plot(kind='bar')
plt.title("Valores faltantes por variable")
plt.show()
```

```python
df['event_type'].value_counts().plot(kind='bar')
plt.title("Distribución de eventos")
plt.show()
```

```python
(df['event_type'].value_counts(normalize=True) * 100).plot(kind='bar')
plt.title("Porcentaje de eventos")
plt.ylabel("Porcentaje (%)")
plt.show()
```
